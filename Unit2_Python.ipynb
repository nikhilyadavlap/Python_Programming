Question 1: Check Voting Eligibility with Nested if

Problem:
Write a program that checks whether a person is at least 18 years old. If they are 18 older, check whether they are a citizen.
Only citizens who are 18 or older are eligible to vote.

```python
age = int(input("Enter a age: "))
citizen = input("Are you a citizen? (yes/no): ")

if age >= 18:
    if citizen == "yes":
        print("You are eligible to vote. ")
    else:
        print("You are not eligible to vote.")
else:
    print("You are underage.")
```

Question 5: Calculate the Sum of Numbers.

Problem:
Write a python program to calculate the sum of all numbers from 1 to 10 using a for loop

```python
total = 0

for i in range(1, 11):
    total += i

print("Sum = ", total)
```

Question 6: Print a multiplication table.

Problem:
Write a Pyhton program that takes a number from the user and prints its multiplication table from 1 to 10.

```python
number = int(input("Enter a number: "))

for i in range(1, 11):
    print(number, "x", i, "=", number * i)
```

Question 7: Print numbers using a While Loop

Problem:
Write a python program to print numbers from 1 to 10 using a while loop.

```python
i = 1

while i <= 10:
    print(i)
    i += 1
```

Question 8: Stop a loop using break

Problem:
Write a python program to print numbers from 1 to 10, but stop the loop when the number becomes 6.

```python
for i in range(1, 11):
    if i == 6:
        break

    print(i)
```

Question 9: Skip a number using continue

Problem:
Write a python program to print numbers from 1 to 10 but skip the number 5.

```python
for i in range(1, 11):
    if i == 5:
        continue

    print(i)
```

Question 10: Print a star pattern using nested loops

Problem:
Write a python program using nested for loops to print the following patter:


```python
for i in range(1, 6):
    for j in range(i):
        print("*", end="")

    print()
```
