Question 3: Add Two Numbers Using a Function

Problem:
Write a function named add() that accepts two numbers and returns their sum.

```python
def add(a , b):
    return a + b

result = add(20, 30)

print("Sum = ", result)
```

Question 4: Calculate the Square

Problem:
Write a function that accepts a number and returns its square.

```python
def square(number):
    return number ** 2

result = square(8)

print("Square =", result)
```

Question 5: Find the Largest of Two Numbers

Problem:
Write a function that accepts two numbers and return the larger number.

```python
def largest(a , b):
    if a > b:
        return a
    else:
        return b

result = largest(45, 72)

print("Largest =", result)
```
