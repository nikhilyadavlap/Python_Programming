Question 20. Grade Calculatore
Write a python program to print the grade based on marks.
. 90-100 --> Grade A
. 75-89 --> Grade B
. 60-74 --> Grade C
. 40-59 --> Grade D
. Below 40 --> Fail

```python
marks = int(input("Enter marks: "))

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
elif marks >= 40:
    print("Grade D")
else:
    print("Fail")
```
